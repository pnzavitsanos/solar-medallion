```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
RAW = ROOT / "data" / "raw" / "v2"
print(RAW)

fs = pd.read_csv(RAW / "fusionsolar_energy_production.csv")
fs.shape
fs.head()
```

```python
so = pd.read_csv(RAW / "solis_energy_production.csv")
so.shape
so.head()
```

```python
print("unique hashes fs",fs["device_hash_id"].nunique())
print("unique hashes so",so["device_hash_id"].nunique())
```

```python
fs_counts = fs["device_hash_id"].value_counts()
so_counts = so["device_hash_id"].value_counts()

summary = pd.DataFrame(
    {
        "fs": [len(fs_counts), len(fs), round(fs_counts.mean()),
               fs_counts.idxmax(), fs_counts.max(),
               fs_counts.idxmin(), fs_counts.min()],
        "so": [len(so_counts), len(so), round(so_counts.mean()),
               so_counts.idxmax(), so_counts.max(),
               so_counts.idxmin(), so_counts.min()],
    },
    index=["number of devices", "total rows", "average rows per device",
           "device with most rows", "most rows",
           "device with least rows", "least rows"],
)
summary
```
